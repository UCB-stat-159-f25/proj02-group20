# Part 2: Simple Text Processing - Tokenization, Lemmatization, Word Frequency, Vectorization 

```python
# imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-dark') 
```

```python
import spacy
from tqdm import tqdm
from collections import Counter

nlp = spacy.load("en_core_web_sm")
```

```python
sou = pd.read_csv('data/SOTU.csv')
sou
```

### Processing Speeches with SpaCy

```python
above_2000 = sou[sou['Year'] >= 2000].copy()
above_2000
```

```python
text_generator = above_2000['Text'].tolist()
docs = list(nlp.pipe(text_generator, disable=["parser", "ner"]))

all_clean_lemmas = []

for doc in tqdm(docs, desc="Extracting Tokens"):

    clean_tokens = [
        token.lemma_
        for token in doc          
        if not token.is_stop       
        and not token.is_punct    
        and not token.is_space   
        and not token.like_num  
        and token.is_alpha
    ]
    all_clean_lemmas.extend(clean_tokens)
```

### Analyze Tokens vs Lemmas

#### Token List

```python
all_clean_tokens = []

for doc in docs:
    clean_tokens = [
        token.text.lower()
        for token in doc
        if not token.is_stop 
        and not token.is_punct 
        and not token.is_space 
        and not token.like_num 
        and token.is_alpha 
    ]
    all_clean_tokens.extend(clean_tokens)

token_list = all_clean_tokens

word_counts_raw = Counter(token_list)
top_20_tokens = word_counts_raw.most_common(20)

top_20_tokens
```

#### Lemma List


```python
# Assuming 'docs' is the list of spaCy Doc objects for speeches >= 2000

all_clean_lemmas = []

# Iterate over all processed documents
for doc in docs:
    # Filter tokens and extract the lemma (base form)
    clean_lemmas = [
        token.lemma_.lower()        # 1. Get the base form (lemma) and make it lowercase
        for token in doc             # 2. Iterate over all tokens
        if not token.is_stop        # 3. Exclude stop words
        and not token.is_punct      # 4. Exclude punctuation
        and not token.is_space      # 5. Exclude white space
        and not token.like_num      # 6. Exclude numbers
        and token.is_alpha          # 7. Only include alphabetical tokens
    ]
    all_clean_lemmas.extend(clean_lemmas)


word_counts_lemma = Counter(all_clean_lemmas)

# Use the .most_common() method to display the top 20 (lemma, count) tuples
top_20_lemmas = word_counts_lemma.most_common(20)

top_20_lemmas
```

#### Token versus Lemma Comparison


The top tokens and the top lemmas differ. The count in lemma frequency seems be larger or equal than in tokens. This is because in lemma frequency it is counting the base words and token counts the exact word. An example of this is shown in "year" and "years". Their token frequency is considered seperately so they are smaller but in lemma frequency the count of "year" is much larger because it considers "years" and "year". "child" only appears in the top 20 of the lemmas list because it includes the count of other words containing "child" such as "children" for example.

### Common Words

#### Common Words per Year Function

```python
def get_most_common_words(df, year, n=25):
    """
    Processes the SOTU speech for a given year and returns
    the most common non-stopword/punctuation lemmas.
    """

    # Step 1: Subset df
    speech_df = df[df['Year'] == year].copy()

    # Step 2: Process the text with spaCy

    doc = nlp(speech_df['Text'].iloc[0], disable=["parser", "ner"])
    
    # Step 3: Get lemmas
    clean_lemmas = [
        token.lemma_.lower()
        for token in doc
        if not token.is_stop      # Filter out common stop words (e.g., 'the', 'a')
        and not token.is_punct    # Filter out punctuation
        and not token.is_space    # Filter out whitespace/newlines
        and not token.like_num    # Filter out numbers/numeric-like tokens
    ]
    
    return Counter(clean_lemmas).most_common(n)
```

```python
#test it on 2024
get_most_common_words(sou, 2024, n=10)
```

#### Compare 2023 to 2017


```python
words_2023 = get_most_common_words(sou, 2023, n=20)
words_2017 = get_most_common_words(sou, 2017, n=20)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA
```

#### Train the Vectorizer and Transform the Data

```python
raw_docs = sou["Text"].to_list()
```

```python
np.random.seed(42)

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(raw_docs)
tfidf_array = tfidf_matrix.toarray()

pca = PCA(n_components=2, random_state=42)
principal_components = pca.fit_transform(tfidf_matrix)
```

#### Plot Speeches

```python
df_pca = pd.DataFrame(data=principal_components, columns=['Principle Component 1', 'Principle Component 2'])
df_pca

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_pca, x='Principle Component 1', y='Principle Component 2', s=60)

plt.title('Plot of Vectorized Speeches Principle Components')
plt.xlabel('Principle Component 1')
plt.ylabel('Principle Component 2')
plt.grid(False)
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


sns.set_style("white")
plt.figure(figsize=(10, 15)) # Adjusted size for better visibility of rows/columns

feature_names = vectorizer.get_feature_names_out()

N = 928 # Adjust this number based on your actual total number of features (tokens)

subset_matrix = tfidf_array[:, ::N]
subset_feature_names = feature_names[::N]

from matplotlib.colors import LogNorm

sns.heatmap(
    data=subset_matrix, # Plot the subset matrix
    norm=LogNorm(vmin=1e-3, vmax=tfidf_array.max()), # Adjust vmin/vmax as needed
    cmap="magma", # Color map similar to the example (magma or viridis can work)
    cbar_kws={'label': 'TF-IDF Score', 'ticks': [1e-3, 1e-2, 1e-1]}
)

plt.title('Vectorized Speeches', fontsize=16)
plt.ylabel('Speech Index', fontsize=14)
plt.xlabel('Vector Index', fontsize=14)


plt.xticks(
    ticks=np.arange(len(subset_feature_names)) + 0.5, # Position ticks in the center of the columns
    labels=[f'{idx}\n{name}' for idx, name in enumerate(subset_feature_names)],
    rotation=90, # Rotate labels for better fit
    fontsize=8
)

num_speeches = subset_matrix.shape[0]
y_ticks = np.arange(0, num_speeches, 10) # Set ticks every 10 speeches
plt.yticks(y_ticks, y_ticks, rotation=0, fontsize=10)

plt.tight_layout()
plt.show()
```

#### Get the TF-IDF value for certain words and documents

```python
word_list = ['year',
 'america',
 'people',
 'american',
 'work',
 'new',
 'job',
 'country',
 'americans',
 'world'] # top ten most common words through whole corpus
```

```python
word_nums = [tfidf.vocabulary_.get(word) for word in word_list] # get each word's index number using the .vocabular_ attributed of vectorizer
```

```python
idf_score = tfidf.idf_[word_nums] # get their IDF score by using .idf_ at the indices from the previous step
```

```python
tf_idf = tfidf_array[0, word_nums] # get the tf_idf score for the first speech
```

```python
pd.DataFrame({"Word": word_list, "IDF Score": idf_score, "TF-IDF Score": tf_idf})
```
